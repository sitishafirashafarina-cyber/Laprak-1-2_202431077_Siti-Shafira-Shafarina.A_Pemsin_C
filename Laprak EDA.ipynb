## Menghitung data kosong

```python
#202431077_Siti Shafira Shafarina.A
df.isna().sum()
```

## Melihat nilai unik kolom kategorik

```python
#202431077_Siti Shafira Shafarina.A
kolom_kategorik = df.select_dtypes(
    include='object'
).columns

for kolom in kolom_kategorik:
    print("Kolom:", kolom)
    print(df[kolom].unique())
    print()
```

## Mengisi data kosong

```python
#202431077_Siti Shafira Shafarina.A

# Mengisi data kosong pada kolom kategorik dengan modus

for kolom in kolom_kategorik:
    df[kolom] = df[kolom].fillna(
        df[kolom].mode()[0]
    )

# Mengisi data kosong pada kolom numerik dengan mean

kolom_numerik = df.select_dtypes(
    include='number'
).columns

for kolom in kolom_numerik:
    df[kolom] = df[kolom].fillna(
        df[kolom].mean()
    )

# Memeriksa kembali data kosong

df.isna().sum()
```

## Menghitung data duplikat

```python
#202431077_Siti Shafira Shafarina.A
df.duplicated().sum()
```

## Menghapus data duplikat

```python
#202431077_Siti Shafira Shafarina.A
df = df.drop_duplicates()

df.info()
```

## Visualisasi grafik batang

```python
#202431077_Siti Shafira Shafarina.A
jumlah_gender = df['gender'].value_counts()

jumlah_gender.plot(kind='bar')

plt.title('Jumlah Siswa Berdasarkan Gender')
plt.xlabel('Gender')
plt.ylabel('Jumlah Siswa')

plt.xticks(rotation=0)
plt.show()
```
