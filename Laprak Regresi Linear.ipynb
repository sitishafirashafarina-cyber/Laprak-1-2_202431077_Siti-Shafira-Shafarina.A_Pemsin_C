## Import Library

```python
# 202431077_Siti Shafira Shafarina Amirullah

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score
```

## Membaca Dataset

```python
# 202431077_Siti Shafira Shafarina Amirullah

df = pd.read_csv('data_gedung_bersih.csv')

print("Data Awal:")
display(df.head())
```

## Melihat Informasi Dataset 

```python
# 202431077_Siti Shafira Shafarina Amirullah

print("Informasi Dataset:")
df.info()
```

## Menghapus Data Kosong

```python
# 202431077_Siti Shafira Shafarina Amirullah

df = df.dropna(subset=['lantai', 'tinggi_m'])

print("Jumlah Data Setelah Menghapus Data Kosong:", len(df))
print("Jumlah Data Kosong:")
print(df[['lantai', 'tinggi_m']].isnull().sum())
```

## Menentukan Fitur (X) dan Target (Y)

```python
# 202431077_Siti Shafira Shafarina Amirullah

X = df[['lantai']]
y = df['tinggi_m']

print("Fitur (X):")
display(X.head())

print("Target (Y):")
display(y.head())
```

## Membagi Data Latih dan Data Uji

```python
# 202431077_Siti Shafira Shafarina Amirullah

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, random_state=0
)

print("Jumlah Data Latih:", len(X_train))
print("Jumlah Data Uji:", len(X_test))
```

## Membuat Model Linear Regression

```python
# 202431077_Siti Shafira Shafarina Amirullah

model = LinearRegression()

print("Model Linear Regression berhasil dibuat.")
```

## Melatih Model

```python
# 202431077_Siti Shafira Shafarina Amirullah

model.fit(X_train, y_train)

print("Model berhasil dilatih.")
```

## Melakukan Prediksi

```python
# 202431077_Siti Shafira Shafarina Amirullah

y_pred = model.predict(X_test)

print("Hasil Prediksi:")
print(y_pred[:10])
```

## Evaluasi Model

```python
# 202431077_Siti Shafira Shafarina Amirullah

mse = mean_squared_error(y_test, y_pred)
r2 = r2_score(y_test, y_pred)

print("Koefisien:", model.coef_[0])
print("Intercept:", model.intercept_)
print("MSE:", mse)
print("R2 Score:", r2)
```

## Persamaan Regresi Linear

```python
# 202431077_Siti Shafira Shafarina Amirullah

a = model.intercept_
b = model.coef_[0]

print("Persamaan Regresi Linear:")
print(f"Y = {a:.4f} + ({b:.4f} X)")
```
